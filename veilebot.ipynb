# veile-bot 動作確認ノートブック

毎朝の AI ダイジェストを生成・確認するための手元用ノート。

**認証情報はここに書かない。** すべて環境変数から読む。


````python
import os

# 必要なら .env 相当をここで設定する（このセルはコミットしないこと）
os.environ.setdefault("VEILE_LLM_ENABLED", "0")  # まずは LLM なしで確認
````

````python
from veile_bot.config import Config
from veile_bot.sources import build_sources, collect

config = Config.from_env()
items = collect(build_sources(config))
len(items), items[:3]
````

````python
from veile_bot.ranking import select
from veile_bot.store import NullStore

selected = select(items, config, NullStore())
for entry in selected:
    print(f"{entry.score:.3f}  {entry.item.title}")
    print("        ", " / ".join(entry.reasons))
````

````python
from veile_bot.pipeline import build_digest
from veile_bot.render import preview

digest, collected = build_digest(config, NullStore())
print(preview(digest))
````
